# Data Preparation - Titanic Dataset

## Objective
The objective of this project is to prepare a public Titanic dataset
for data science analysis by cleaning missing values, handling data
types, removing duplicates, and documenting assumptions.

```python
import pandas as pd
import numpy as np
df = pd.read_csv("C:/Users/chili/OneDrive/Desktop/data-preparation/data/Titanic-Dataset.csv")
print("Dataset loaded successfully!")
```

## Missing Value Analysis

Before cleaning the dataset, missing values are identified in each column.

```python
missing_values = df.isnull().sum()
print(missing_values)
```

```python
print("Number of duplicate rows:", df.duplicated().sum())
```

### Handling Missing Age Values

Age is a numerical variable. Missing age values are replaced with the
median age because the median is less affected by extreme values than
the mean.

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
```

### Handling Missing Embarked Values

Embarked is a categorical variable. Missing values are replaced with
the mode because it represents the most frequently occurring category.

```python
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
```

### Handling Cabin Values

The Cabin column contains a large proportion of missing values.
Instead of assigning potentially incorrect cabin information, the
column is removed from the prepared dataset.

```python
df = df.drop(columns=["Cabin"])
```

```python
df = df.drop_duplicates()

print("Duplicate rows after cleaning:", df.duplicated().sum())
```

## Data Type Handling

Numerical columns are converted to appropriate numeric types and
categorical columns are stored as categorical data where appropriate.

```python
df["PassengerId"] = pd.to_numeric(df["PassengerId"], errors="coerce")
df["Survived"] = pd.to_numeric(df["Survived"], errors="coerce")
df["Pclass"] = pd.to_numeric(df["Pclass"], errors="coerce")
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
df["Fare"] = pd.to_numeric(df["Fare"], errors="coerce")

df["Sex"] = df["Sex"].astype("category")
df["Embarked"] = df["Embarked"].astype("category")
```

## Verification

The dataset is checked again after cleaning to ensure that missing
values and duplicate records have been handled.

```python
print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nData types:")
print(df.dtypes)
```
